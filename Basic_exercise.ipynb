# WAP to ask user enter a basic salary, DAmount & HRA amount. Calculate total pay.

```python
basic_salary=eval(input('Enter basic salary'))
DA= eval(input('Enter DA amount'))
HRA=eval(input('Enter HRA amount' ))
Totalpay=basic_salary+DA+HRA
print('Total supply is:',Totalpay)
```

# WAP to ask the user to enter salary, tax percentage. Calculate how much tax user need to pay.


```python
salary=eval(input('Enter salary'))
tax_percentage=eval(input('Enter tax percentage'))
tax_amount=salary*(tax_percentage/100)
print('The total salary is ', tax_amount)
```

# WAP that asks 3 numbers(3 separate input). Create total & average that hold sum & average of 3 numbers & print them.

```python
num_1=eval(input('Enter the first number'))
num_2=eval(input('Enter the second number'))
num_3=eval(input('Enter the third number'))
total=num_1+num_2+num_3
average=total/3
print('The sum of three number is',total)
print('The average of three number is',average)
```
